```python
def calculate_sales(price, quantity):
    return price * quantity


def calculate_discount(sales, discount_rate):
    return sales * discount_rate


def calculate_profit(net_sales, cost):
    return net_sales - cost



product = "Laptop"
price = 55000
quantity = 3
discount_rate = 0.10
cost = 120000


total_sales = calculate_sales(price, quantity)
discount = calculate_discount(total_sales, discount_rate)
net_sales = total_sales - discount
profit = calculate_profit(net_sales, cost)


print("Product:", product)
print("Total Sales:", total_sales)
print("Discount:", discount)
print("Net Sales:", net_sales)
print("Profit:", profit)
```

```python
import pandas as pd


data = {
    "Product": [
        "Laptop", "Mouse", "Keyboard",
        "Monitor", "Mouse", "Printer"
    ],
    "Price": [
        "55000", "800", None,
        "12000", "800", "6000"
    ],
    "Quantity": [
        3, 10, 5, 4, 10, None
    ]
}

df_dirty = pd.DataFrame(data)

print("Original Dataset:")
print(df_dirty)


print("\nMissing Values:")
print(df_dirty.isnull().sum())


print("\nDuplicate Records:")
print(df_dirty.duplicated().sum())


df_clean = df_dirty.drop_duplicates().copy()

df_clean["Price"] = pd.to_numeric(
    df_clean["Price"], errors="coerce"
)


df_clean["Price"] = df_clean["Price"].fillna(
    df_clean["Price"].median()
)

df_clean["Quantity"] = df_clean["Quantity"].fillna(
    df_clean["Quantity"].median()
)


df_clean["Quantity"] = df_clean["Quantity"].astype(int)


print("\nCleaned Dataset:")
print(df_clean)


print("\nMissing Values After Cleaning:")
print(df_clean.isnull().sum())
```

```python
import numpy as np


prices = np.array(df_clean["Price"])

print("Product Prices:", prices)


mean_price = np.mean(prices)
median_price = np.median(prices)
max_price = np.max(prices)
min_price = np.min(prices)
std_price = np.std(prices)
total_price = np.sum(prices)


print("\nStatistical Analysis:")
print("Mean Price:", mean_price)
print("Median Price:", median_price)
print("Maximum Price:", max_price)
print("Minimum Price:", min_price)
print("Standard Deviation:", round(std_price, 2))
print("Sum of Prices:", total_price)
```

```python
import pandas as pd


df_clean["Revenue"] = df_clean["Price"] * df_clean["Quantity"]


total_revenue = df_clean["Revenue"].sum()

total_units = df_clean["Quantity"].sum()

average_revenue = df_clean["Revenue"].mean()

highest_revenue_product = df_clean.loc[
    df_clean["Revenue"].idxmax(), "Product"
]


print("BUSINESS KPI ANALYSIS")
print("Total Revenue: Rs.", total_revenue)
print("Total Units Sold:", total_units)
print("Average Revenue per Product: Rs.", round(average_revenue, 2))
print("Highest Revenue Product:", highest_revenue_product)

print("\nUpdated Sales Dataset:")
print(df_clean)
```

```python
import matplotlib.pyplot as plt


plt.figure(figsize=(9, 5))

plt.bar(
    df_clean["Product"],
    df_clean["Revenue"],
    color="lavender",
    edgecolor="black"
)


plt.title("Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue (Rs.)")


plt.xticks(rotation=30)


plt.grid(axis="y", linestyle="--", alpha=0.5)


plt.tight_layout()


plt.show()
```
