```python
import json
import pandas as pd

with open("users.json", "r") as file:
    data = json.load(file)

df = pd.json_normalize(data)

df = df[
    [
        "id",
        "name",
        "username",
        "email",
        "address.city",
        "phone",
        "website",
        "company.name"
    ]
]

df.columns = [
    "User_ID",
    "Name",
    "Username",
    "Email",
    "City",
    "Phone",
    "Website",
    "Company"
]

df.to_csv("users_data.csv", index=False)

print("Total Records:", len(df))
print("CSV file created successfully!")

display(df)
```
